```python
import plotly.graph_objects as go
import ipywidgets as widgets
from plotly.subplots import make_subplots
from IPython.display import display
import vtk
import numpy as np
from vtk.util.numpy_support import vtk_to_numpy

# Loading
reader = vtk.vtkXMLImageDataReader()
reader.SetFileName('mixture.vti')
reader.Update()  # Update() in VTK to execute the read.

image_data = reader.GetOutput()

# Extracting
point_data = image_data.GetPointData()
# Get the active scalars.
vtk_array = point_data.GetScalars() 

# Convert VTK array to a standard NumPy array
values = vtk_to_numpy(vtk_array)

# 3. Generate X, Y, Z coordinates
dims = image_data.GetDimensions()
spacing = image_data.GetSpacing()
origin = image_data.GetOrigin()

# Creating 1D arrays for each axis
x_axis = origin[0] + np.arange(dims[0]) * spacing[0]
y_axis = origin[1] + np.arange(dims[1]) * spacing[1]
z_axis = origin[2] + np.arange(dims[2]) * spacing[2]

# Put Z first so that when flattened, X changes fastest, perfectly matching VTK's internal memory layout.
Z, Y, X = np.meshgrid(z_axis, y_axis, x_axis, indexing='ij')

# Flatten them into 1D arrays to match your original PyVista output
x = X.flatten()
y = Y.flatten()
z = Z.flatten()

# 4. Calculate bounds and print
min_val = float(np.min(values))
max_val = float(np.max(values))

print(f'Data range: {min_val:.4f} to {max_val:.4f}')
```

```python
# Create initial isosurface and histogram
x_list = x.tolist()
y_list = y.tolist()
z_list = z.tolist()
values_list = values.tolist()

iso_trace = go.Isosurface(
    x=x_list,
    y=y_list,
    z=z_list,
    value=values_list,
    isomin=0.0,
    isomax=0.0,
    surface_count=1,
    colorscale='plasma', # Using same as assignment
    cmin=min_val,
    cmax=max_val,
    showscale=False,
    caps=dict(x_show=False, y_show=False, z_show=False)
)

hist_trace = go.Histogram(
    x=values_list,
    nbinsx=29,  # Using same bins as shown in assignment
    marker_color='royalblue'
)

fig = go.FigureWidget(make_subplots(
    rows=1, cols=2, 
    specs=[[{'type': 'scene'}, {'type': 'xy'}]],
)) 
fig.add_trace(iso_trace, row=1, col=1)
fig.add_trace(hist_trace, row=1, col=2)

fig.update_layout(
    height=500,
    width=900,
    showlegend=False,
    margin=dict(l=0, r=0, b=0, t=30)
)
fig.update_yaxes(title_text="Frequency", row=1, col=2)
fig.update_xaxes(title_text="Vortex scalar values", row=1, col=2)
fig.update_scenes(
    xaxis_title='x', yaxis_title='y', zaxis_title='z',
    camera=dict(eye=dict(x=1.5, y=1.5, z=1.5)),
    row=1, col=1
)
```

```python

slider = widgets.FloatSlider(
    value=0.0,
    min=min_val,
    max=max_val,
    step=(max_val - min_val) / 200,
    description='Isoval:',
    continuous_update=False, # Update plot only when sliding stops
    readout_format='.2f',
    layout=widgets.Layout(width='400px')
)  # Slider

reset_button = widgets.Button(description='Reset') # Reset Button

def update_plots(change):
    isoval = change['new']
    
    with fig.batch_update():
        # Update isosurface
        fig.data[0].isomin = isoval
        fig.data[0].isomax = isoval
        
        # Update histogram
        mask = (values >= (isoval - 0.25)) & (values <= (isoval + 0.25))
        fig.data[1].x = values[mask].tolist()

slider.observe(update_plots, names='value')

def reset_plots(b):
    slider.value = 0.0
    with fig.batch_update():
        # Isosurface reset to 0.0
        fig.data[0].isomin = 0.0
        fig.data[0].isomax = 0.0
        
        # Histogram reset to full data
        fig.data[1].x = values_list

        fig.update_scenes(
            xaxis_title='x', yaxis_title='y', zaxis_title='z',
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.5)),
            row=1, col=1
        ) # Camera Reset back to original state

reset_button.on_click(reset_plots)

# Render
controls = widgets.HBox([slider, reset_button])
display(widgets.VBox([controls, fig]))
```
